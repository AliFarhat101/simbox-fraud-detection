### Load the dataset into DuckDB for easier querying

```python
from pathlib import Path
import duckdb

PROJECT_ROOT = Path.cwd().parent.parent
DB_PATH = PROJECT_ROOT / "database" / "simbox.duckdb"

print("Using DB:", DB_PATH)

con = duckdb.connect(str(DB_PATH))
```

### Create tables from csv

```python
file1 = PROJECT_ROOT / "data" / "raw_cdr" / "cdr_6.8million_30kMSISDNS_v3.csv"
file2 = PROJECT_ROOT / "data" / "raw_cdr" / "ali_sleiman_msisdns_cdr.csv"
file3 = PROJECT_ROOT / "data" / "profiles" / "all_profiles.parquet"

con.execute(f"""
CREATE TABLE IF NOT EXISTS cdr_all AS
SELECT * FROM read_csv_auto('{file1}')
""")

con.execute(f"""
CREATE TABLE IF NOT EXISTS cdr_fraud AS
SELECT * FROM read_csv_auto('{file2}')
""")

con.execute(f"""
CREATE TABLE IF NOT EXISTS profiles AS
SELECT * FROM '{file3}'
""")
```

Sanity checks

```python
con.execute("SHOW TABLES").fetchall()
```

Close database connection

```python
con.close()
```
